```python
import pandas as pd
import numpy as np 
import seaborn as sns
```

```python
df=pd.read_csv("C:\\Users\\cl601_02\\Downloads\\Iris_data (1).csv")
df.head()
```

```python
df.columns=["Sepal_length","Sepal_width","Petal_Length","Petal_width","Class"]
df.head()
```

```python
df.Class.unique()
```

```python
lable={'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2}
y=[lable[a] for a in df.iloc[:,-1]]
```

```python
y
```

```python
X=df.iloc[:,:-1]
X
```

```python
colormap=np.array(['red','lime','black'])
import matplotlib.pyplot as plt
plt.title("Real")
plt.scatter(X.Petal_Length,X.Petal_width,c=colormap[y])
```

```python
from sklearn.cluster import KMeans
model=KMeans(n_clusters=3,random_state=0)
model.fit(X)
KMeans_labels=model.predict(X)
```

```python
KMeans_labels
```

```python
plt.title("Kmeans")
plt.scatter(X.Petal_Length,X.Petal_width,c=colormap[KMeans_labels])
```

```python
# sum from i 1 to n (modulus ( xi - cl(i))^2)
#xi= datapoint
#cl= cluster centeroid
```

```python
from sklearn.metrics import silhouette_score,confusion_matrix,accuracy_score
print("Inertia:",model.inertia_)
print("silhouette_score:",silhouette_score(X,KMeans_labels))
print('Accuracy:',accuracy_score(y,KMeans_labels))
print("Confusion Matrix:\n",confusion_matrix(y,KMeans_labels))
```

```python
from sklearn.mixture import GaussianMixture
em=GaussianMixture(n_components=3,random_state=0).fit(X)
y_cluster_em=em.predict(X)
y_cluster_em
```

```python
plt.title("Gaussian")
plt.scatter(X.Petal_Length,X.Petal_width,c=colormap[y_cluster_em])
```

```python
print("silhouette_score:",silhouette_score(X,y_cluster_em))
print('Accuracy:',accuracy_score(y,y_cluster_em))
print("Confusion Matrix:\n",confusion_matrix(y,y_cluster_em))
```

```python
plt.figure(figsize=(24, 6)) 

plt.subplot(1, 3, 1)
plt.title("Real")
plt.scatter(X.Petal_Length,X.Petal_width,c=colormap[y])

plt.subplot(1, 3, 2)
plt.title("Kmeans")
plt.scatter(X.Petal_Length,X.Petal_width,c=colormap[KMeans_labels])

plt.subplot(1, 3, 3)
plt.title("Gaussian")
plt.scatter(X.Petal_Length,X.Petal_width,c=colormap[y_cluster_em])

plt.show()
```
