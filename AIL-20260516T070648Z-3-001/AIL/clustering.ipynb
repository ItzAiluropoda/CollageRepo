```python
import pandas as pd
import numpy as np 
```

```python
df=pd.read_csv("C:\\Users\\cl601_17\\Downloads\\Datasets\\Dataset\\Iris_data.csv")
df.head()
```

```python
df.columns=["Sepal_length","Sepal_width","Petal_Length","Petal_width","Class"]
df.head()
```

```python
df.Class.unique()
```

```python
lable={'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2}
y=[lable[a] for a in df.iloc[:,-1]]
```

```python
y
```

```python
X=df.iloc[:,:-1]
X
```

```python
colormap=np.array(['red','lime','black'])
import matplotlib.pyplot as plt
plt.title("Real")
plt.scatter(X.Petal_Length,X.Petal_width,c=colormap[y])
```

```python
from sklearn.cluster import KMeans
model=KMeans(n_clusters=3)
model.fit(X)
KMeans_labels=model.predict(X)
```

```python
KMeans_labels
```

```python
plt.title("Kmeans")
plt.scatter(X.Petal_Length,X.Petal_width,c=colormap[KMeans_labels])
```

```python
# sum from i 1 to n (modulus ( xi - cl(i))^2)
#xi= datapoint
#cl= cluster centeroid
```

```python
from sklearn.metrics import silhouette_score
inertia=model.inertia_
silhouette_avg=silhouette_score(X,KMeans_labels)

print("Inertia:",inertia)
print("silhouette_score:",silhouette_avg)
```
