```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

NORMAL SCALER

```python
frame=pd.DataFrame(data=[[5,5,1],[10,2,0],[2,0,4],[0,10,5]],columns=['X','Y',"Z"])
frame
```

```python
plt.scatter(frame.X,frame.Y)
plt.axvline(x=0,c='red',label='x=0')
plt.axhline(y=0,c='yellow',label='y=0')
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
Scaler=MinMaxScaler()
Min_Max_scaled=Scaler.fit_transform(frame)
```

```python
scaled_frame=pd.DataFrame(Min_Max_scaled,columns=['X','Y','Z'])
scaled_frame
```

```python
plt.scatter(scaled_frame.X,scaled_frame.Y)
plt.axvline(x=0,c='red',label='x=0')
plt.axhline(y=0,c='yellow',label='y=0')
plt.show()
```

```python
fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(projection='3d')
ax.scatter(scaled_frame.X,scaled_frame.Y,scaled_frame.Z)
#ax.axvline(x=0,c='red',label='x=0')
#ax.axhline(y=0,c='yellow',label='y=0')
plt.show()
```

```python
plt.scatter(frame.X,frame.Y)
plt.axvline(x=0,c='red',label='x=0')
plt.axhline(y=0,c='yellow',label='y=0')
plt.show()

plt.scatter(scaled_frame.X,scaled_frame.Y)
plt.axvline(x=0,c='red',label='x=0')
plt.axhline(y=0,c='yellow',label='y=0')
plt.show()
```

```python
df=pd.read_csv("C:\\Users\\cl601_07\\Documents\\24070127070\\Dataset\\Scaling_cars.csv")
df
```

```python
import matplotlib.pyplot as plt

plt.scatter(df.Volume,df.Weight)
plt.axvline(x=0,c='red',label='x=0')
plt.axhline(y=0,c='yellow',label='y=0')
plt.show()
```

MIN MAX

```python
from sklearn.preprocessing import MinMaxScaler
min_max_scaler=MinMaxScaler()
min_max_scaled=min_max_scaler.fit_transform(df[['Volume','Weight']])
```

```python
min_max_frame=pd.DataFrame(min_max_scaled,columns=['Volume','Weight'])
min_max_frame
```

```python
plt.scatter(df.Volume,df.Weight)
plt.axvline(x=0,c='red',label='x=0')
plt.axhline(y=0,c='yellow',label='y=0')
plt.show()

plt.scatter(min_max_frame.Volume,min_max_frame.Weight)
plt.axvline(x=0,c='red',label='x=0')
plt.axhline(y=0,c='yellow',label='y=0')
plt.show()
```

STANDARD SCALE

```python
from sklearn.preprocessing import StandardScaler
STD_scaler=StandardScaler()
STD_scaled=STD_scaler.fit_transform(df[['Volume','Weight']])
```

```python
STD_frame=pd.DataFrame(STD_scaled,columns=['Volume','Weight'])
STD_frame
```

```python
plt.scatter(df.Volume,df.Weight)
plt.axvline(x=0,c='red',label='x=0')
plt.axhline(y=0,c='yellow',label='y=0')
plt.show()

plt.scatter(min_max_frame.Volume,min_max_frame.Weight)
plt.axvline(x=0,c='red',label='x=0')
plt.axhline(y=0,c='yellow',label='y=0')
plt.show()

plt.scatter(STD_frame.Volume,STD_frame.Weight)
plt.axvline(x=0,c='red',label='x=0')
plt.axhline(y=0,c='yellow',label='y=0')
plt.show()
```
