```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(18,5))

plt.subplot(1,3,1)
sns.countplot(x="Fuel_Type", data=df, color="blue",
              order=df["Fuel_Type"].value_counts().index)
plt.title("Fuel_Type")

plt.subplot(1,3,2)
sns.countplot(x="Transmission", data=df, color="pink",
              order=df["Transmission"].value_counts().index)
plt.title("Transmission")

plt.subplot(1,3,3)
sns.countplot(x="Owner_Type", data=df, color="lightgreen",
              order=df["Owner_Type"].value_counts().index)
plt.title("Owner_Type")

plt.tight_layout()
plt.show()

```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(18,5))

plt.subplot(1,3,1)
df["Fuel_Type"].value_counts().plot.pie(autopct="%1.1f%%")
plt.title("Fuel Type")
plt.ylabel("")

plt.subplot(1,3,2)
df["Transmission"].value_counts().plot.pie(autopct="%1.1f%%")
plt.title("Transmission")
plt.ylabel("")

plt.subplot(1,3,3)
df["Owner_Type"].value_counts().plot.pie(autopct="%1.1f%%")
plt.title("Owner_Type")
plt.ylabel("")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(18,10))
pie_kwargs={
    "autopct":'%1.1f%%',
    "pctdistance":0.85,
    "labeldistance":1.1,
    "wedgeprops":{"linewidth":5,"edgecolor":"white"}

}

#pie chart for fuel type
plt.subplot(2,3,1)
plt.pie(df["Fuel_Type"].value_counts(),labels=df["Fuel_Type"].value_counts().index,colors=["skyblue","lightgreen","lightcoral"],**pie_kwargs)
plt.title("Fuel Type Distribution")
plt.gca().axis("equal")

#pie chart for transmission
plt.subplot(2,3,2)
plt.pie(df["Transmission"].value_counts(),labels=df["Transmission"].value_counts().index,colors=["blue","red","green"],**pie_kwargs)
plt.title("Transmission Distribution")
plt.gca().axis("equal")

plt.subplot(2,3,3)
plt.pie(df["Owner_Type"].value_counts(),labels=df["Owner_Type"].value_counts().index,colors=["lightgreen","red","green"],**pie_kwargs)
plt.title("Owner_Type Distribution")
plt.gca().axis("equal")
```
