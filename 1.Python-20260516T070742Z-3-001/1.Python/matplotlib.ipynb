```python
#creates blank canvas
fig=plt.figure()


axes1=fig.add_axes([0 , 0 , 1 , 1]) #big fig
axes2=fig.add_axes([0.2 , 0.2 , 0.5 , 0.5]) #small fig

#larger figure Axes 1
axes1.plot(a,b)

#use set_to add to the axes figure
axes1.set_xlabel("X label")
axes1.set_ylabel("Y label")
axes1.set_title("Big figure")

#insert figure axes 2
axes2.plot(a,b)
axes2.set_title("Small figure")

plt.show()
```

```python
#creates blank canvas
fig=plt.figure()


axes1=fig.add_axes([0 , 0 , 1 , 1]) #big fig
axes2=fig.add_axes([0.2 , 0.2 , 0.5 , 0.5]) #small fig

#larger figure Axes 1
axes1.plot(a,b)

#use set_to add to the axes figure
axes1.set_xlabel("X label")
axes1.set_ylabel("Y label")
axes1.set_title("Big figure")

#insert figure axes 2
axes2.plot(a,b)
axes2.set_title("Zoomed IN")
axes2.set_xlim(8,10)
axes2.set_ylim(4000,10000)
axes2.set_xlabel("X")
axes2.set_ylabel("Y")

plt.show()
```

```python
#creates blank canvas
fig=plt.figure()


axes1=fig.add_axes([0 , 0 , 1 , 1]) #big fig
axes2=fig.add_axes([0.25 , 0.25 , 0.25 , 0.25]) #small fig
axes3=fig.add_axes([1,1,0.25,0.25]) # start at top right corner

#larger figure Axes 1
axes1.plot(a,b)

#use set_to add to the axes figure
axes1.set_xlabel("X label")
axes1.set_ylabel("Y label")
axes1.set_title("Big figure")

#insert figure axes 2
axes2.plot(a,b)
axes2.set_title("Zoomed IN")
axes2.set_xlim(8,10)
axes2.set_ylim(4000,10000)
axes2.set_xlabel("X")
axes2.set_ylabel("Y")

#add axes3
axes3.plot(a,b)
axes3.set_xlabel("X axis")
axes3.set_ylabel("Y axis")
axes3.set_title("corner figure")

plt.show()
```

```python
fig,axes=plt.subplots(nrows=2,ncols=2,figsize=(12,8))
axes[0][0].plot(a,b)
axes[0][0].set_title('0 0 title')

axes[1][1].plot(b,a)
axes[1][1].set_title('1 1 title')
axes[1][1].set_xlabel('1 1 X label')

axes[0][1].plot(x,y)
axes[1][0].plot(x,y)

fig.suptitle("Figure level",fontsize=16)
fig.subplots_adjust(left=None,
                    bottom=None,
                    right=None,
                    top=None,
                    wspace=0.9,
                    hspace=0.1)

plt.show()
```
