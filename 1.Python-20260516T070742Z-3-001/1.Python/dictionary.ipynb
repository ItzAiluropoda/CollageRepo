```python
n=int(input("Enter no of entries:"))
rec={}
for i in range(0,n):
    name=str(input("Enter name:"))
    marks=float(input("Enter % marks:"))
    if marks<0 or marks>100:
        print("Invalid marks,Entry removed")
    else:
        rec[name]=marks
        print("Entry added")
print(rec)
```

```python
d1={100:"durga",200:"ravi",300:"shiva"}
d1[400]="cake"
print(d1)
del d1[100]
print(d1)
d1.clear()
print(d1)
del d1
print(d1)
```

```python
d2=dict()
print(d2)
d3=dict({100:"ravi",200:"beans"})
print(d3)
d4=dict([(100,"shiva"),(200,"cake")])
print(d4)
```

```python
#python 3
if 400 in d4:
    print(True)
else:
    print(False)

#python 2
print(d4.has_key(400))
```

```python
d5={100:"durga",200:"ravi",300:"shiva"}
print(d5.keys())
for k in d5.keys():
    print(k)
print("\n")
print(d5.values())
for k in d5.values():
    print(k)
print("\n")
for k,v in d5.items():
    print(k,"--",v)
```
